```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
data = {
    "Name": ["Arun", "Priya", "Kumar", "arun", "Divya", None, "Rahul", "Priya"],
    "Age": [22, 25, 30, 22, None, 28, -5, 25],
    "Department": ["IT", "HR", "Sales", "it", "Finance", "IT", "Sales", "HR"],
    "Salary": [25000, 30000, 40000, 25000, 35000, None, 20000, 30000],
    "City": ["Chennai", "Mumbai", "Delhi", "chennai", "Pune", "Chennai", None, "Mumbai"]
}

df = pd.DataFrame(data)

df.to_csv("messy_employee_data.csv", index=False)

df
```

```python
print("Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nDataset Information:")
df.info()
```

```python
# Remove duplicate records
df = df.drop_duplicates()

# Fill missing Name and City values
df['Name'] = df['Name'].fillna('Unknown')
df['City'] = df['City'].fillna('Unknown')

# Fill missing Age and Salary with median
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Salary'] = df['Salary'].fillna(df['Salary'].median())

# Correct invalid age
df.loc[df['Age'] < 0, 'Age'] = df['Age'].median()

# Standardize text values
df['Name'] = df['Name'].str.title()
df['Department'] = df['Department'].str.title()
df['City'] = df['City'].str.title()

print("Data cleaning completed!")
display(df)
```

```python
print("========== CLEANING REPORT ==========")

print("\nFinal Dataset Shape:")
print(df.shape)

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nDuplicate Records After Cleaning:")
print(df.duplicated().sum())

print("\nAge Statistics:")
print(df['Age'].describe())

print("\nSalary Statistics:")
print(df['Salary'].describe())
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['Age'], bins=5, kde=True)
plt.title("Age Distribution of Employees")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.show()

plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='Department', y='Salary')
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.show()
```

```python
print("========== DATA CLEANING SUMMARY ==========")

print("Original Dataset Rows: 8")
print("Cleaned Dataset Rows:", len(df))

print("\nMissing Values After Cleaning:")
print(df.isnull().sum().sum())

print("\nDuplicate Rows After Cleaning:")
print(df.duplicated().sum())

print("\nFinal Columns:")
print(df.columns.tolist())

print("\nFinal Cleaned Dataset:")
display(df)
```

```python
## Conclusion

Data cleaning was successfully performed on the employee dataset.

- Missing values were handled.
- Duplicate records were removed.
- Invalid age values were corrected.
- Text values were standardized.
- Data visualizations were created.
- The cleaned dataset was saved as a CSV file.

This process improves data quality and makes the dataset suitable for further analysis.
```
